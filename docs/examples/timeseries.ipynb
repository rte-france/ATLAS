# Timeseries Class Examples

## 1. Creating Timeseries Objects


```python
from atlas import Timeseries
from datetime import datetime
import polars as pl
import pandas as pd

### Empty Timeseries
ts = Timeseries()
print(f"Empty timeseries:\n{ts}")

### From Dictionary
data_dict = {
    "time": [datetime(2025, 1, 1), datetime(2025, 1, 2), datetime(2025, 1, 3)],
    "value": [10.5, 11.2, 9.8],
}
ts_from_dict = Timeseries(data_dict)
print(f"From dictionary:\n{ts_from_dict}")

### From Polars DataFrame
df = pl.DataFrame(
    {
        "time": [datetime(2025, 1, 1), datetime(2025, 1, 2), datetime(2025, 1, 3)],
        "value": [10.5, 11.2, 9.8],
    }
)
ts_from_polars = Timeseries(df)
print(f"From Polars DataFrame:\n{ts_from_polars}")

### From Pandas DataFrame
pd_df = pd.DataFrame(
    {
        "timestamp": [datetime(2025, 1, 1), datetime(2025, 1, 2), datetime(2025, 1, 3)],
        "measurement": [10.5, 11.2, 9.8],
    }
)
ts_from_pandas = Timeseries(pd_df)
print(f"From pandas DataFrame:\n{ts_from_pandas}")
```


## 2. Setting and Getting Values


```python
ts = Timeseries()
# Add values
ts.set_value("2025-01-01 12:00:00", 10.5)
ts.set_value("2025-01-01 13:00:00", 11.2)
ts.set_value("2025-01-01 14:00:00", 9.8)
print(f"After setting values:\n{ts}")

# Get a value
value = ts.get_value("2025-01-01 13:00:00")
print(f"Value at 2025-01-01 12:00:00: {value}")
```


## 3. Time Series Operations


```python
# Create two timeseries
ts1 = Timeseries(
    {
        "time": [datetime(2025, 1, 1), datetime(2025, 1, 2), datetime(2025, 1, 3)],
        "value": [10.0, 20.0, 30.0],
    }
)

ts2 = Timeseries(
    {
        "time": [datetime(2025, 1, 1), datetime(2025, 1, 2), datetime(2025, 1, 4)],
        "value": [1.0, 2.0, 4.0],
    }
)

# Addition
ts_sum = ts1 + ts2
print(f"Addition result:\n{ts_sum}")

# Multiplication
ts_scaled = ts1 * ts2
print(f"Multiplied result:\n{ts_scaled}")
```


## 4. Resampling and Aggregation


```python
# Create hourly data
dates = Timeseries.generate_datetimes(
    start="2025-01-01 00:00:00",
    end="2025-01-03 23:00:00",
    freq="1h",
    timezone="UTC",
    date_format="YYYY-MM-DD HH:mm:ss",
)
values = [float(i % 24) for i in range(len(dates))]
hourly_ts = Timeseries({"time": dates, "value": values})

# Group by day and calculate mean
daily_ts = hourly_ts.groupby("1d", agg="mean")
print(f"Daily aggregation (mean):\n{daily_ts}")
```


## 5. Upsampling and Interpolation


```python
# Create a daily time series
daily_data = {
    "time": [datetime(2025, 1, 1), datetime(2025, 1, 2), datetime(2025, 1, 3)],
    "value": [10.0, 20.0, 30.0],
}
daily_ts = Timeseries(daily_data)

# Upsample to hourly with linear interpolation
hourly_ts = daily_ts.upsample("1h")
print(f"Upsampled to hourly (linear):\n{hourly_ts}")
```


## 6. Filtering


```python
# Create a week of data
week_dates = Timeseries.generate_datetimes(
    start="2025-01-01", end="2025-01-07", freq="1d", timezone="UTC", date_format="YYYY-MM-DD"
)
week_values = [float(i * 10) for i in range(len(week_dates))]
week_ts = Timeseries({"time": week_dates, "value": week_values})

# Filter for a specific date
filtered_ts = week_ts.filter("2025-01-03", date_format="YYYY-MM-DD", inplace=False)
print(f"Filtered for 2025-01-03:\n{filtered_ts}")
```


## 7. Handling Missing Values


```python
# Create a time series with missing values
missing_data = {
    "time": [
        datetime(2025, 1, 1),
        datetime(2025, 1, 2),
        datetime(2025, 1, 3),
        datetime(2025, 1, 4),
        datetime(2025, 1, 5),
    ],
    "value": [10.0, None, 30.0, None, 50.0],
}
missing_ts = Timeseries(missing_data)
print(f"Time series with missing values:\n{missing_ts}")

# Linear interpolation
linear_interp = missing_ts.interpolate(inplace=False)
print(f"After linear interpolation:\n{linear_interp}")
```

## 8. Visualization


```python
# Create a time series with some data
viz_dates = Timeseries.generate_datetimes(
    start="2025-01-01", end="2025-01-31", freq="1d", timezone="UTC", date_format="YYYY-MM-DD"
)

import math

viz_values = [100 + 30 * math.sin(i * 0.3) for i in range(len(viz_dates))]
viz_ts = Timeseries({"time": viz_dates, "value": viz_values})

# Create a plot
fig = viz_ts.plot(
    title="Monthly Data Visualization",
    height=500,
    width=800,
    line_color="darkblue",
    line_shape="spline",
)
fig.show()
```


## 9. Exporting Data


```python
to_file_ts = Timeseries(
    {
        "time": [datetime(2025, 1, 1), datetime(2025, 1, 2), datetime(2025, 1, 3)],
        "value": [10.0, 20.0, 30.0],
    }
)

# Export to different formats

to_file_ts.to_file("timeseries_data.csv", file_format="csv")
to_file_ts.to_file("timeseries_data.parquet", file_format="parquet")
to_file_ts.to_file("timeseries_data.pickle", file_format="pickle")
```
