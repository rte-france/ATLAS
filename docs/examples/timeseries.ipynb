```python
from atlas import Timeseries

import polars as pl
import pandas as pd
import datetime

# Parameters
start_time = datetime.datetime(2023, 1, 1, 0, 0)
n = 10  # number of time steps
freq_minutes = 60  # time delta in minutes

# Generate time series data
time_series = [start_time + datetime.timedelta(minutes=freq_minutes * i) for i in range(n)]
values = [i * 1.5 for i in range(n)]


ts = Timeseries(pd.DataFrame({
    "time": time_series,
    "value": values,
}))

ts.get_data()
```
